```python
from collections import deque, Counter

# --- REQUIRED INPUTS ---
SEED_NUM = 7 
LOCAL_HAZARD = "TYPHOON"
CONTROL_NUM = max(3, SEED_NUM) # 7

# --- INITIALIZATION ---
# Deque with maxlen = 5 (Strict memory limit)
rolling_buffer = deque(maxlen=5)

burst = [ 
    (CONTROL_NUM, "WEATHER"),  
    (CONTROL_NUM + 1, "TRAFFIC"),  
    (CONTROL_NUM + 2, "WEATHER"),  
    (CONTROL_NUM + 3, "WEATHER"),  
    (CONTROL_NUM + 4, LOCAL_HAZARD),  
    (CONTROL_NUM + 5, "WEATHER"),  
    (CONTROL_NUM + 6, "TRAFFIC") 
]

# Data Expansion: Adding (CONTROL_NUM + 2) = 9 custom alerts
# Start timestamps from CONTROL_NUM + 7
for i in range(7, 16):
    category = "TRAFFIC" if i % 2 == 0 else "FIRE"
    burst.append((CONTROL_NUM + i, category))

# --- ALGORITHM (The Sliding Window) ---
append_check_weather = "N"
append_check_hazard = "N"

for alert in burst:
    timestamp, category = alert
    
    # Extract categories currently in deque
    current_categories = [item[1] for item in rolling_buffer]
    counts = Counter(current_categories)
    
    # Spam Rule: Reject if category exists 2 or more times
    if counts[category] < 2:
        rolling_buffer.append(alert)
        # Tracking specific check-questions for the table
        if timestamp == CONTROL_NUM + 3 and category == "WEATHER":
            append_check_weather = "Y"
        if category == LOCAL_HAZARD:
            append_check_hazard = "Y"

# Final Calculations
final_sum = sum(alert[0] for alert in rolling_buffer)

# --- OUTPUTS ---
print(f"LOCAL_HAZARD Used: {LOCAL_HAZARD}")
print(f"Total Burst Size: {len(burst)}")
print(f"Did 'WEATHER' at {CONTROL_NUM + 3} append?: {append_check_weather}")
print(f"Did {LOCAL_HAZARD} append?: {append_check_hazard}")
print(f"Final Deque Length: {len(rolling_buffer)}")
print(f"Final Sum of Timestamps: {final_sum}")
print(f"Exact Final Deque Output: {list(rolling_buffer)}")
```
